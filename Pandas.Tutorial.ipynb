```python
#Cleaning data 
df = pd.read_csv("raw_data.csv")

df.isnull()
df.isna()

df.isnull().sum()   #to find missing value in data 

df.dropna()   #drop all null value from row
df.dropna(axis=1) #drop all null value from row

df.fillna(100)    #fill all mising value from 0


#perticular any column value we can clean and set to a perticular value
age_mean = df["age"].mean()

cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean()
cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)


#forward fill
df.ffill()
#backword fill
df.bfill()
```

```python
#Handle Duplicate Data
df.duplicated()
df["country"].duplicated()
df[["country","gender"]].duplicated()

#remove duplicate value from data

#(inplace=True) when we replace original data from file
df.drop_duplicates() 
```

```python
#Data Types

df.dtypes
df2 = df.copy()

df2 = df2.fillna(0)

df2["age"] = df2["age"].astype("int64").copy()
df2.dtypes
```

```python
#Cleaning data handle string
df["gender"].str.lower()
df["gender"].str.upper()
df["gender"].str.capitalize()

#suhailazhr@gmail.com
#df["name"].str.split("@")  if we have split name from gmail

#if the value exist we can check throuh countains
df["country"].str.contains("india", case=False)
```

```python
#Transform of data (feature Engineering)
df2 = df.copy()

#def logic:
#    if (income < 50_000):



df2["tax"] = df2["income"].apply(lambda x : "20%" if x >= 50000 else "10%" )

gender_map = {"Male" : "M" , "Female" : "F" , "Unknown" : "U"}
df2["gender"] = df2["gender"].map(gender_map)

df2 = df2.assign(new_income = df["income"] * 1.1 )

#replace of any data from original data 
df2["country"] = df2["country"].replace("USA","United State")

#rename of the data 
df2.columns = ["ID","Name","Age","Country","Gender","Income","Tax","New_Income"]
df2.rename(columns={"Income":"Salary"})
df2.rename(index={1:"First"})

#Sort the data means assending order
df2.sort_values("Income")
df2 = df2.fillna(50)

#Sort the data means desending order
df2.sort_values("Income" , ascending=False)


#sort_data = ["Income","Age"]
sorted_df = df2.sort_values(["Income","Age"])

#Sort index value 
sorted_df.sort_index()

#reset index value 
sorted_df.reset_index()
sorted_df.reset_index(drop=True)  #we can drop the original index value from data 

#ranking method in pandas
#sorted_df["Ranking"] = sorted_df["Income"].rank( ascending=False, method="dense")
sorted_df[["ID","Name","Age","Country","Gender","Income","New_Income","Tax"]]

```
